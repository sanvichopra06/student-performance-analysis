```python
marks = [78, 45, 32, 90, 67, 38, 81]
pass_count =0
fail_count=0
for m in marks:
    if m>=75:
        print(m, "Distinction")
        pass_count +=1
    elif m>=40: 
        print(m, "Pass")
        pass_count +=1
    else:
        print(m, "Fail")
        fail_count +=1
print()
print("Pass", pass_count)
print("Fail", fail_count)
print("Total", pass_count+fail_count)
```

```python
marks= [88, 34, 76, 45, 23, 90, 61, 39, 50]
cp=0
cf=0

for m in marks:
    if m>=75:
        print(m ,"Distinction")
        cp +=1
    elif m>=40:
        print(m ,"Pass")
        cp +=1
    else:
        print(m, "Fail")
        cf +=1
print()
print("Total= ", cp+cf)
print("Pass= ",cp)
print("Fail= ",cf)
```

```python
marks = [88, 34, -5, 76, 120, 45, 23, 90, 61, 39, 50]
cp=0
cf=0
cd=0
ci=0
for m in marks:
    if m<0 or m>100:
        print(m ,"Invalid")
        ci +=1
    elif m>=75:
        print(m ,"Distinction")
        cd +=1
    elif m>=40:
        print(m ,"Pass")
        cp +=1
    else:
        print(m, "Fail")
        cf +=1
print("Total= ", ci+cf+cp+cd)
print("Valid= ", cf+cd+cp)
print("Invalid= ", ci)
print("Distinction= ",cd)
print("Pass= ",cp)
print("Fail= ",cf)
        
```

```python
marks= [88, 76, 45, 23, 90, 61, 39, 50]
cp=0
cf=0
total = len(marks)
for m in marks:
    if m>=75:
        cp +=1
    elif m>=40:
        cp +=1
    else:
        cf +=1
print()
print("Pass % = ", (cp/total*100))
print("Fail % = ",(cf/total)*100)
```

Computing Average

```python
marks= [80,90,76,54,32,75,90]
totalmarks=0

for m in marks:
    totalmarks+=m

average= totalmarks/len(marks)
print("Average: ",average)
```
