Student Performance Analysis 

Analyze student marks to understand overall performance, pass rates,
and identify potential areas of concern.

The dataset contains marks of students from a single exam.
Marks are expected to be in the range 0–100.

```python
marks = [55, 78, 92, 33, 47, 61, 29, 85]
```

We first ensure that all values are valid marks.

```python
passcount = 0
failcount = 0
totalmarks = 0

for m in marks:
    if m < 0 or m > 100:
        print(m, "Invalid")
    else:
        totalmarks += m
        if m >= 40:
            passcount += 1
        else:
            failcount += 1
```

```python
passrate = (passcount / len(marks) * 100)
averagemarks = totalmarks / len(marks)

print("Total students:", len(marks))
print("Pass %:", passrate)
print("Average marks:", averagemarks)
```

Insights:
1) Majority of students passed the exam.
2) Average marks indicate moderate overall performance.
3) Some students are close to the pass threshold, suggesting
  a need for targeted academic support.
